# 字典

字典储存一些一一对应的 "键" (key) 和 "值" (value). 可以通过键来快速查找值.

一般来说, 键是一些**不可变**的常量, 比如字符串、数字等, 而值几乎可以是任何东西: 数字, 字符串, 列表, 字典等等都可以.

## 基本用法

使用 `my_dict = {key_1: value_1, key_2: value_2, ...}` 来表示一个字典; 使用 `my_dict[key]` 来查找对应的 `value`:

```python
my_dict = {"key_1": "value_1", "key_2": "value_2",}

print(my_dict["key_1"])
```

## 添加、修改、删除

添加和修改一个字典的键值对很简单: 如果想要添加 `"key_3": "value_3"`, 直接使用 `my_dict["key_3"] = "value_3"` 即可; 如果想修改某个键对应的值, 直接赋值 `my_dict["key_1"] = "new_value_1"` 即可.

```python
my_dict["key_3"] = "value_3"
my_dict["key_1"] = "new_value_1"

print(my_dict)
```

若想要删除某个键值对, 直接使用 `del my_dict["key_2"]` 即可:

```python
del my_dict["key_2"]

print(my_dict)
```

## 遍历字典

可以使用 `for` 循环遍历字典: 

 - 若想要同时遍历一个字典的键和值, 可以通过 `for key, value in my_dict.items()` 来遍历;
 - 若想要单独遍历字典的键或值, 可以通过 `for key in my_dict.keys()` (或 `for keys in my_dict`) 和 `for value in my_dict.values()` 来遍历. 

这三个方法返回 "视图" 对象, 可以被 `for` 循环遍历. 

注意: 在遍历字典的时候, 不要修改字典:

 - 若添加或删除条目, 则可能会引发 `RuntimeError` 或漏掉条目;
 - 若更新已有键的值, 程序会按照新的值视图继续遍历.

```python
my_dict = {
    "key_1": "value_1", 
    "key_2": "value_2", 
    "key_3": "value_3", 
    }

for key, value in my_dict.items():
    print("The value of key " + key + " is " + value)


print("\nAll the keys in my_dict are:")
for key in my_dict.keys():
    print("    " + key)

print("\nAll the values in my_dict are:")
for value in my_dict.values():
    print("    " + value)
```
