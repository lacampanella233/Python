## 基本概念

列表由一些元素有序排列而成, 用方括号表示, 元素用逗号隔开.

## 访问列表

列表可以直接打印. 

```python
MyList = [1,2,3,4]
print(MyList)
```

在列表名的后面直接加上某个被方括号括起来的整数 $n$, 即表示这个列表的索引为 $n$ 的元素. 注意: 索引从 $0$ 开始. 

$n$ 可以为负数. 在不超过索引范围的情况下, 索引 $-n$ 表示列表的倒数第 $n$ 个元素.

```python
Mylist = [1,2,3,4]
print(Mylist[0])
print(Mylist[-1])
print(Mylist[0] + Mylist[-1])
```

## 操作列表的元素

 - `append(element)`: 添加新的元素 `element` 到列表*末尾*.
 - `insert(n,element)`: 将 `element` 插入到列表中, 使得它的索引为 $n$. 如果索引 $n$ 超出了边界, 则会将元素插入到列表开头或结尾.
 - `del`: 删除给定列表的给定索引的元素. 
 - `pop(n)`: 删除列表的索引为 $n$ 的元素, 并且返回此元素的值. 若索引没有给出, 则默认为 $-1$, 即默认弹出最后一个元素. 
 - `remove(element)`: 删除列表的*第一个*等于 `element` 的元素. 

```python
my_list = [1,2,4,5]
print("my_list now is " + str(my_list) + ".")
my_list.append(6)
print("my_list now is " + str(my_list) + " after append 6.")
my_list.insert(2,3)
print("my_list now is " + str(my_list) + " after insert 3 at index 2.")

del my_list[2]
print("my_list now is " + str(my_list) + " after delete the element at index 2.")
poped_element = my_list.pop()
print("my_list now is " + str(my_list) + " and the popped element is " + str(poped_element) + ".")
my_list.remove(1)
print("my_list now is " + str(my_list) + " after remove the first element with value 1.")
```

## 重排列表

 - `sort()`: 升序重排列表并*修改*原列表. 若括号内有 `reverse=True` 则为降序排列.
 - `sorted(list)`: 返回升序重排的 `list`, 同时*不修改*原列表. 若括号内有 `reverse=True` 则为降序排列.
 - `reverse()`: 将列表元素反序并*修改*原列表. 
 - `len(list)`: 返回 `list` 的长度.

```python
my_list = [1,4,2,8,5,7]
print("The reversely sorted list is " + str(sorted(my_list,reverse=True)) + ", \nbut the original list is still " + str(my_list) + ".")
my_list.sort()
print("Now the original list is " + str(my_list) + " after sorting.")

my_list.reverse()
print("Now the list is " + str(my_list) + " after reverting.")
print("The length of the list is " + str(len(my_list)) + ".")
```
