## `for` 循环

```python
for element in my_list:
    some code here
```

`for` 循环会将下方相同缩进的代码重复执行, 并且 `element` 变量会遍历列表 `my_list`, 遍历完成之后会结束循环执行后面的代码. 缩进不正确会引发 `IndentationError`. 遗漏冒号会引发 `SyntaxError`. 

````python
my_list = [1,2,3,4]
for number in my_list:
    print(number)
print("Done!")
````

## 数值列表

### `range()`

`range()` 可以为 `for` 循环创建等差数列构成的列表. 有如下几种用法:
 - `range(n)`: 遍历 $0,1,\dots,n-1$, 其中 $n\in\mathbb{Z}$(否则出现 `TypeError`, 错误地将浮点数填入应当填整数的地方). 若 $n\le 0$, 则不进行循环.
 - `range(a,b)`: 遍历 $a,a+1,\dots,b-1$, 其中 $a,b\in\mathbb{Z}$. 若 $a\ge b$, 则不进行循环.
 - `range(a,b,d)`: 若 $d>0$ 则遍历 $a,a+d,\dots,a+\lceil(b-a)/d-1\rceil d$, 即从 $a$ 开始, 每次增加 $d$, 若下一步大于等于 $b$ 则到此为止. $d<0$ 的情况同理. $a,b,d\in\mathbb{Z}$, 且 $d\neq 0$(否则会出现 `ValueError`).

````python
my_list = []
for i in range(5):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(-5,5):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(-5,5,2):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(5,-5,-3):
    my_list.append(i)
print(my_list)
````

### 统计计算

`min()`, `max()`, `sum()` 可以对数字列表进行求最小/大值以及求和. 

````python
numbers = [1,2,3,4]
print(min(numbers))
print(max(numbers))
print(sum(numbers))
````

## 列表解析

这是一种很方便的语法, 使用如下:

````python
numbers = [1,2,4,5,7,8]

squares = [x**2 for x in numbers]
cubes = [x**3 for x in range(1,11)]

print(squares)
print(cubes)
````

## 切片

在列表中使用冒号可以访问列表的一部分连续的元素. 使用 `[a:b]` 以访问以索引为 `[a]` 的元素开始、索引为 `[b-1]` 的元素结束的切片. **注意:切片不包含以 `[b]` 为索引的元素.** 

若 $a$ 留空, 则默认从一个元素开始, 即默认 $a=0$; 同理, 若 $b$ 留空 ,则默认以最后一个元素结束. 

````python
my_list = [1,2,3,4,5]
print(my_list[1:3])
print(my_list[2:])
print(my_list[:3])
````
