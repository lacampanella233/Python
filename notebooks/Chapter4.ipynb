## `for` 循环

```python
for element in my_list:
    some code here
```

`for` 循环会将下方相同缩进的代码重复执行, 并且 `element` 变量会遍历列表 `my_list`, 遍历完成之后会结束循环执行后面的代码. 缩进不正确会引发 `IndentationError`. 遗漏冒号会引发 `SyntaxError`. 

````python
my_list = [1,2,3,4]
for number in my_list:
    print(number)
print("Done!")
````

## 数值列表

### `range()`

`range()` 可以指定 `for` 循环的遍历范围为一个等差数列. 有如下几种用法:
 - `range(n)`: 遍历 $0,1,\dots,n-1$, 其中 $n\in\mathbb{Z}$(否则出现 `TypeError`, 错误地将浮点数填入应当填整数的地方). 若 $n\le 0$, 则不进行循环.
 - `range(a,b)`: 遍历 $a,a+1,\dots,b-1$, 其中 $a,b\in\mathbb{Z}$. 若 $a\ge b$, 则不进行循环.
 - `range(a,b,d)`: 若 $d>0$ 则遍历 $a,a+d,\dots,a+\lceil(b-a)/d-1\rceil d$, 即从 $a$ 开始, 每次增加 $d$, 若下一步大于等于 $b$ 则到此为止. $d<0$ 的情况同理. $a,b,d\in\mathbb{Z}$, 且 $d\neq 0$(否则会出现 `ValueError`).
 - `list(range(...))`: 返回一个列表, 其内容为 `range(...)` 所指定的等差数列.

````python
my_list = []
for i in range(5):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(-5,5):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(-5,5,2):
    my_list.append(i)
print(my_list)

my_list = []
for i in range(5,-5,-3):
    my_list.append(i)
print(my_list)
````

### 统计计算

`min()`, `max()`, `sum()` 可以对数字列表进行求最小/大值以及求和. 

````python
numbers = [1,2,3,4]
print(min(numbers))
print(max(numbers))
print(sum(numbers))
````

## 切片

在列表中使用冒号可以访问列表的一部分连续的元素. 使用 `[a:b]` 以访问以索引为 `[a]` 的元素开始、索引为 `[b-1]` 的元素结束的切片. **注意:切片不包含以 `[b]` 为索引的元素.** 

若 $a$ 留空, 则默认从一个元素开始, 即默认 $a=0$; 同理, 若 $b$ 留空 ,则默认以最后一个元素结束. $a$ 与 $b$ 可以为负整数.

````python
my_list = [1,2,3,4,5]
print(my_list[1:3])
print(my_list[2:])
print(my_list[:3])
print(my_list[:])
print(my_list[-3:])
````

若索引为 `[a]` 的元素在索引为 `[b-1]` 的元素之后, 则返回空列表.

````python
my_list = [1,2,3,4,5]
print(my_list[3:2])
````

切片可以被遍历:

````python
my_list = [1,2,3,4,5]
for i in my_list[1:3]:
    print(i)
````

### 复制列表

使用以下两种方法复制列表有根本性的差别:

```python
new_list = old_list		# 方法 1
new_list = old_list[:]	# 方法 2
```

方法 1 将 `new_list` 与 `old_list` 这两个变量指向了同一个列表对象. 若将该变量进行*原地修改*, 则两个变量都会变化; 若将该变量*重新赋值*, 则不会影响另一个对象.

方法 2 创建了一个新的列表, 其内容与 `old_list` 所指向的列表相同, 并且将变量 `new_list` 指向此列表.

````python
old_list = [1]
new_list = old_list		# 按照方法 1 “复制”列表

# 对两个变量进行原地修改:

old_list.append(2)
print(new_list)			# 验证 `new_list` 是否被修改

new_list.append(3)
print(old_list)			# 验证 `old_list` 是否被修改

old_list.append(4)
print(new_list)			# 再次验证 `new_list` 是否被修改

# 对两个变量进行重新赋值:

old_list = [1]
print(old_list)         # `old_list` 被更新
print(new_list)         # `new_list` 没有随之更新
````

````python
old_list = [1]
new_list = old_list[:]	# 按照方法 2 复制列表

old_list.append(2)
print(new_list)			# 验证 `new_list` 是否被修改

new_list.append(3)
print(old_list)			# 验证 `old_list` 是否被修改

old_list.append(4)
print(new_list)			# 再次验证 `new_list` 是否被修改
````

## 元组

元组可以简单理解为 "不能修改的列表". 

元组通过逗号来标识. 注意: 若要创建只有一个元素的元组, 应当多加一个逗号, 即 `(1,)`. 

元组的元素不能被再次赋值, 否则返回 `TypeError`; 不过, 可以通过重新定义元组的方式来修改元组的值. 

````python
my_tuple = (1,2,3)
print(my_tuple)

my_tuple = 4,5,6
print(my_tuple)

my_tuple = 1,
print(my_tuple)
````
