# `if` 语句

## 条件测试

 - 使用 `==` 或 `!=` 来检查两个对象是否相等或不相等. 
 - 使用 `is` 来检查两个变量是否引用同一个对象. 
 - 对于数字, 也可以使用 `<`, `>`, `<=` 或 `>=` 来比较大小.
 - 使用 `element in list` 来检查 `element` 是否在 `list` 中, 使用 `element not in list` 来检查 `element` 是否不在 `list` 中. 
 - 条件测试返回布尔值 `True` 或 `False`.

## 布尔值的运算

使用 `and` 与 `or` 对两个布尔值做“与”运算或“或”运算. 

## `if` 语句

可以使用 `if`, `elif` 或 `else` 来组织条件判断. 如果某个分支满足了条件, 在执行这个分支内的程序之后, 将**不再判断与运行同一个语句内后面的分支**. 

在使用 `else` 中, 若可以明确希望 `else` 分支执行的条件, 则最好使用 `elif 条件` 来代替.

与 `for` 循环一样, 使用 `if` 语句时, 必须有明确且正确的缩进. 不要忘记冒号.

可以使用的句式: 

```py
if...
if...else...
if...elif...elif...         # (elif 一个或多个均可)
if...elif...elif...else...  # (elif 一个或多个均可)
```
