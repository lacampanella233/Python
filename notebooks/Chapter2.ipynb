## 字符串操作

 - `title()`: 将每个单词的首字母转换为大写, 并且将其余字母转换为小写.

```python
message = 'this is a title'
print(message.title())

message = 'THIS IS A TITLE'
print(message.title())

message = 'ThIs Is A tItLe'
print(message.title())
```

 - 合并: 将两个字符串用`+`连接以合并. 
 - 空白: `\t` 是制表符, `\n` 是换行符.
 - 删除空白: 有 `strip()`, `lstrip()` 和 `rstrip()` 三种, 分别用于删去字符串两端、左端与右端的空白.

```python
message = '\n1\n\t2\n3\n'

print('Output1: ' + message)
print('Output2: ' + message.strip())
print('Output3: ' + message.lstrip())
print('Output4: ' + message.rstrip())
```

## 数字

Python 3 中的整数 `int` 与浮点数 `float` 可以混合运算. 

 - `str()`: 将数字转换为字符串. 例如, 如果要将数字与字符串结合, 应当使用 `str()` 将数字转换为字符串.

```python
integer = 1
float_number = 1.5

print('integer = ' + str(integer))
print('float_number = ' + str(float_number))
```

注意: 浮点数的运算可能不完全精确.

```python
print(0.1+0.2)
```
